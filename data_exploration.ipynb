## Missing Values (overview)

```python
df = df.replace([
    "", " ", "-", "N/A", "n/a", "NA", "na",
    "None", "none", "null", "NULL"
], np.nan)
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage (%)": (df.isnull().sum() / len(df) * 100).round(2)
})

print(missing_summary.sort_values("Missing Percentage (%)", ascending=False))
```

## Train-validation-test split

```python
from sklearn.model_selection import train_test_split

unique_ids = df["id"].unique() # split w/ ids

train_ids, temp_ids = train_test_split(
    unique_ids,
    test_size=0.30,
    random_state=42
)

val_ids, test_ids = train_test_split(
    temp_ids,
    test_size=0.50,
    random_state=42
)

train_df = df[df["id"].isin(train_ids)]
val_df = df[df["id"].isin(val_ids)]
test_df = df[df["id"].isin(test_ids)]

X_train = train_df.drop(columns=["id", "target_sport"])
X_val = val_df.drop(columns=["id", "target_sport"])
X_test = test_df.drop(columns=["id", "target_sport"])

y_train = train_df["target_sport"]
y_val = val_df["target_sport"]
y_test = test_df["target_sport"]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_val shape:", X_val.shape)
print("y_val shape:", y_val.shape)

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

# check no overlap between train, val, and test sets

train_ids = set(train_ids)
val_ids = set(val_ids)
test_ids = set(test_ids)

print("Train & validation overlap:", train_ids & val_ids)
print("Train & test overlap:", train_ids & test_ids)
print("Validation & test overlap:", val_ids & test_ids)
```

## Class Balance

```python
for col in train_df.select_dtypes(
    include=["object", "category", "bool"]
).columns:
    print(f"\n--- {col} ---")

    percentages = (
        train_df[col]
        .value_counts(normalize=True, dropna=False)
        .mul(100)
        .round(2)
    )

    print(percentages)
```

```python
multi_category_cols = [
    "play_season",
    "associated_music",
    "play_location"
]  

for col in multi_category_cols:
    print(f"\n--- {col} ---")

    percentages = (
        train_df[col]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
    )

    print(percentages)
```

## Feature Distribution

```python
import matplotlib.pyplot as plt

numerical_cols = train_df.select_dtypes(include=["int64", "float64"]).columns

train_df[numerical_cols].hist(
    figsize=(15, 10),
    bins=20,
    edgecolor="black"
)

plt.tight_layout()
plt.show()
```

```python
multi_category_cols = [
    "play_season",
    "associated_music",
    "play_location"
]

for col in multi_category_cols:
    category_counts = (
        train_df[col]
        .dropna()
        .str.split(",")
        .explode()
        .str.strip()
        .value_counts()
    )

    plt.figure(figsize=(8, 5))
    plt.bar(category_counts.index, category_counts.values)

    plt.title(f"Distribution of {col}")
    plt.xlabel("Category")
    plt.ylabel("Number of responses")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
agree_disagree = [
    "strategy_importance",
    "injury_risk_comfort",
    "individual_brilliance"
]

for col in agree_disagree:
    counts = train_df[col].value_counts()

    plt.figure(figsize=(8, 4))
    plt.bar(counts.index.astype(str), counts.values)

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha="right")

    plt.tight_layout()
    plt.show()
```
